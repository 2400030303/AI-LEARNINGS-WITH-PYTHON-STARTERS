Why NumPy?

NumPy arrays are designed for numerical computation and allow us to perform operations efficiently without writing loops for every calculation.

```python
# python list
numbers=[10,20,30,40]
print(numbers*2)
```

```python
#Numpy
import numpy as np
numbers=np.array([10,20,30,40])
print(numbers*2)
```

s designed for numerical operations.

You can do:

arr + 5
arr * 2
arr / 10
arr ** 2

without manually looping.

This idea will become very important when we reach vectorization.

```python
import numpy as np
arr=np.array([2,4,6,8,10])
print(arr)
print(arr+5)
print(arr*3)
```

```python
import numpy as np
arr=np.array([10,20,30,40,50])
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
#shape → structure/dimensions of the array
#ndim → number of dimensions
#size → total number of elements
#dtype → data type of elements
```

```python
import numpy as np
arr=np.array([
    [10,20,30],
    [40,50,60]
])
print(arr)
print(arr.size)
print(arr.shape)
print(arr.ndim)
print(arr.dtype)
```

```python
import numpy as np
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(arr)
print(arr[0,0])
print(arr[1,1])
print(arr[1,2])
print(arr[0])
print(arr[1])
print(arr[:,2])
print(arr[0,1:])
print(arr[1,:2])
#arr[row_start:row_stop, col_start:col_stop]
```

arr[0, 0]       # specific element
arr[0]          # entire row
arr[:, 2]       # entire column
arr[0, 1:]      # part of a row
arr[1, :2]      # part of a row

arr[row_start:row_stop, col_start:col_stop]  # 2D slicing

```python
#Reshaping
import numpy as np
arr=np.array([10,20,30,40,50,60])
arr2=arr.reshape(2,3)
print(arr2)
print(arr2.shape)
```

```python
import numpy as np
print(np.zeros(5))
print(np.ones((2,3)))
print(np.arange(1,11))
print(np.arange(2,10,2))
```

```python
# Array Opeartions
import numpy as np
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a**2)
```

```python
#  7️⃣ Aggregation Functions

#Aggregation means taking many values and producing a summary value.

import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.std(arr))
```

```python
#axis=0 → operate DOWN the rows
#axis=0 → column-wise result
#axis=1 → row-wise result
import numpy as np
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(np.sum(arr,axis=0))
print(np.sum(arr,axis=1))
print(np.mean(arr,axis=0))
print(np.mean(arr,axis=1))

```

```python
#Boolean Masking & Filtering
import numpy as np
arr=np.array([10,25,30,45,50])
print(arr[arr>30])
print(arr[arr<25])
print(arr[(arr>=20)&(arr<=40)])
```

```python
# Broadcasting
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr + 5)
print(arr * 2)

values = np.array([1, 2, 3])
print(arr + values)
```
